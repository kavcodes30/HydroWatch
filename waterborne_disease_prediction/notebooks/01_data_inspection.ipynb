```python
import sys, os
sys.path.insert(0, os.path.abspath('..' + os.sep + 'src'))
import pandas as pd, numpy as np
pd.set_option('display.width', 200)
from data_preprocessing import load_raw
chem, phys = load_raw('../data/raw')
print('chemical', chem.shape, '| physical', phys.shape)
```

```python
# time coverage: the filenames say 2021_2025 - check what is actually there
dt = pd.to_datetime(chem['Data Acquisition Time'], format='%d-%m-%Y %H:%M', errors='coerce')
print(dt.dt.year.value_counts().sort_index())
print('range', dt.min(), '->', dt.max())
```

```python
# missingness of the analytes the project cares about
cols = ['Potential of Hydrogen (pH)','Dissolved oxygen (mg/L)','Total Dissolved Solids (mg/L)','Amonia N (mgN/L)','Nitrate N (mgN/L)']
for c in cols:
    print(f'{c:38s} {pd.to_numeric(chem[c].replace("", np.nan), errors="coerce").notna().mean():.1%}')
```

```python
# the decisive check: does turbidity ever co-occur with dissolved oxygen?
sm = pd.read_csv('../data/processed/station_month_clean.csv')
for a in ['ph','electrical_conductivity','turbidity','tds','ammonia_n','nitrate_n']:
    both = (sm[a].notna() & sm['dissolved_oxygen'].notna()).sum()
    print(f'{a:26s} non-null {sm[a].notna().sum():5d} | with DO {both:5d}')
```
